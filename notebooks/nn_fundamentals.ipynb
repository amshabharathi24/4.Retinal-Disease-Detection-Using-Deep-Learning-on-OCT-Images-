```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# ==========================================
# 1. Activation Functions & Derivatives
# ==========================================
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(a):
    return a * (1 - a)

def tanh(z):
    return np.tanh(z)

def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return (z > 0).astype(float)

# Plot Sigmoid, Tanh, and ReLU
z_vals = np.linspace(-5, 5, 200)
plt.figure(figsize=(10, 3.5))
for idx, (name, fn) in enumerate([("Sigmoid", sigmoid), ("Tanh", tanh), ("ReLU", relu)], 1):
    plt.subplot(1, 3, idx)
    plt.plot(z_vals, fn(z_vals), linewidth=2)
    plt.title(f"{name} Activation")
    plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ==========================================
# 2. Single Perceptron (AND Logic Gate)
# ==========================================
X_and = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([[0], [0], [0], [1]])

weights = np.random.randn(2, 1) * 0.5
bias = np.zeros((1, 1))
lr = 0.5

for epoch in range(500):
    z = np.dot(X_and, weights) + bias
    y_pred = sigmoid(z)
    error = y_pred - y_and
    d_z = error * sigmoid_derivative(y_pred)
    weights -= lr * np.dot(X_and.T, d_z)
    bias -= lr * np.sum(d_z, axis=0, keepdims=True)

print("--- Single Perceptron (AND Gate) Predictions ---")
print(np.round(sigmoid(np.dot(X_and, weights) + bias), 3))

# ==========================================
# 3. 2-Layer Neural Network (XOR Non-Linear Problem)
#    Demonstrating Forward + Backpropagation
# ==========================================
X_xor = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_xor = np.array([[0], [1], [1], [0]])

W1 = np.random.randn(2, 4) * 0.5
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1) * 0.5
b2 = np.zeros((1, 1))
lr = 0.5
losses = []

for epoch in range(3000):
    # Forward Pass
    z1 = np.dot(X_xor, W1) + b1
    a1 = tanh(z1)
    z2 = np.dot(a1, W2) + b2
    y_hat = sigmoid(z2)

    # Binary Cross-Entropy Loss
    loss = -np.mean(y_xor * np.log(y_hat + 1e-8) + (1 - y_xor) * np.log(1 - y_hat + 1e-8))
    losses.append(loss)

    # Backward Pass (Chain Rule)
    dz2 = y_hat - y_xor
    dW2 = np.dot(a1.T, dz2) / len(X_xor)
    db2 = np.mean(dz2, axis=0, keepdims=True)

    da1 = np.dot(dz2, W2.T)
    dz1 = da1 * (1 - a1 ** 2)  # derivative of tanh
    dW1 = np.dot(X_xor.T, dz1) / len(X_xor)
    db1 = np.mean(dz1, axis=0, keepdims=True)

    # Gradient Descent Weight Update
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

print("\n--- 2-Layer MLP (XOR Gate) Predictions ---")
print(np.round(y_hat, 3))

plt.figure(figsize=(6, 3))
plt.plot(losses, color="darkorange", linewidth=2)
plt.title("Backpropagation Training Loss Curve (XOR)")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
